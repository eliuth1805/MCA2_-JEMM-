```
import numpy as np
import matplotlib.pyplot as plt

# SOLO valores válidos
x1 = np.linspace(-10,-2,500)
x2 = np.linspace(2,10,500)

y1 = np.sqrt(x1**2 - 4)
y2 = -np.sqrt(x1**2 - 4)

y3 = np.sqrt(x2**2 - 4)
y4 = -np.sqrt(x2**2 - 4)

plt.figure(figsize=(7,7))

plt.plot(x1,y1,linewidth=2)
plt.plot(x1,y2,linewidth=2)

plt.plot(x2,y3,linewidth=2)
plt.plot(x2,y4,linewidth=2)

plt.axhline(0)
plt.axvline(0)

plt.title(r'Hipérbola $x^2-y^2=4$')

plt.xlabel('x')
plt.ylabel('y')

plt.grid(True)

plt.show()

print("Es una hipérbola centrada en el origen.")
print("Abre hacia la derecha e izquierda sobre el eje x.")
print("Vértices: (2,0) y (-2,0)")
print("Asíntotas: y = x  y  y = -x")
```

```
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(0, np.pi, 100)

x = 2 * np.outer(np.cos(u), np.sin(v))
y = 5 * np.outer(np.sin(u), np.sin(v)) + 4
z = 3 * np.outer(np.ones(np.size(u)), np.cos(v)) - 1

ax.plot_surface(x, y, z)

ax.set_title('Elipsoide')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()

print("Es un elipsoide centrado en (0,4,-1)")
print("Semiejes:")
print("- 2 en dirección x")
print("- 5 en dirección y")
print("- 3 en dirección z")
```

```
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(-2, 2, 100)

u, v = np.meshgrid(u, v)

x = np.sqrt(2) * np.cosh(v) * np.cos(u)
y = (2/3) * np.cosh(v) * np.sin(u)
z = np.sinh(v)

ax.plot_surface(x, y, z)

ax.set_title('Hiperboloide de una hoja')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()

print("Es un hiperboloide de una hoja.")
print("La superficie se abre a lo largo del eje z.")
print("Tiene forma parecida a una torre de enfriamiento.")
print("Las secciones horizontales forman elipses.")
```

```
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

x = np.linspace(-5,5,100)
y = np.linspace(-5,5,100)

x, y = np.meshgrid(x, y)

z = x**2 - 2*y**2

ax.plot_surface(x, y, z)

ax.set_title(r'Paraboloide hiperbólico $z=x^2-2y^2$')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()

print("Es un paraboloide hiperbólico.")
print("Tiene forma de silla de montar.")
print("La superficie se curva hacia arriba en x")
print("y hacia abajo en y.")
print("Las curvas de nivel forman hipérbolas.")
```

```
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

x = np.linspace(-6,6,100)
y = np.linspace(-6,6,100)

x, y = np.meshgrid(x, y)

z = x**2 + 3*y**2 + 2*x - 9*y

ax.plot_surface(x, y, z)

ax.set_title('Paraboloide elíptico')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()

print("Es un paraboloide elíptico.")
print("La superficie abre hacia arriba.")
print("Tiene un mínimo en (-1, 3/2).")
print("Las curvas de nivel forman elipses.")
```

```
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

t = np.linspace(0,5,500)

x = 5*t
y = 10*np.sin(t)
z = -4.9*t**2 + 10*t + 2

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x,y,z)

ax.set_title("Trayectoria 3D del jugador")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

plt.show()

print("Representa el movimiento de un jugador en un videojuego.")
print("La trayectoria combina avance, movimiento lateral y salto.")
```

```
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

t = np.linspace(0,2,500)

x = np.cos(3*t)
y = np.sin(3*t)
z = 2*t**(5/2)

# Punto t=1
t0 = 1

x0 = np.cos(3*t0)
y0 = np.sin(3*t0)
z0 = 2*t0**(5/2)

# Vector tangente
vx = -3*np.sin(3*t0)
vy = 3*np.cos(3*t0)
vz = 5*t0**(3/2)

s = np.linspace(-1,1,100)

xt = x0 + vx*s
yt = y0 + vy*s
zt = z0 + vz*s

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x,y,z,label='Curva')
ax.plot(xt,yt,zt,label='Recta tangente')

ax.set_title("Recta tangente a la curva")

ax.legend()

plt.show()

print("Se muestra la curva y su recta tangente en t=1.")
```

```
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5,5,20)
y = np.linspace(-5,5,20)

X,Y = np.meshgrid(x,y)

# Gradiente de f(x,y)=x^2+y^2
U = 2*X
V = 2*Y

plt.figure(figsize=(8,8))

plt.quiver(X,Y,U,V)

plt.title("Campo gradiente")

plt.xlabel("x")
plt.ylabel("y")

plt.grid(True)

plt.show()

print("Las flechas representan la dirección de máximo crecimiento.")
```
